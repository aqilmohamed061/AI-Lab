```python
from itertools import product

def evaluate_kb(P, Q, R):
    # Knowledge base:
    # Q->P
    # P->NOT Q
    # Q OR R
    s1 = (not Q) or P
    s2 = (not P) or (not Q)
    s3 = Q or R
    return s1 and s2 and s3

def evaluate_query(query, P, Q, R):
    if query == 1:
        return R
    elif query == 2:
        return (not R) or P
    elif query == 3:
        return (not Q) or R

def tt_entails(query):
    # Generate all possible truth assignments
    for P, Q, R in product([False, True], repeat=3):
        kb = evaluate_kb(P, Q, R)
        # Check only models where KB is true
        if kb:
            result = evaluate_query(query, P, Q, R)
            # Counterexample: KB true, query false
            if not result:
                return False
    return True

def display_truth_table():
    print("P Q R | Q->P P->~Q Q|R | KB")
    print("-------------------------------")
    for P, Q, R in product([False, True], repeat=3):
        s1 = (not Q) or P
        s2 = (not P) or (not Q)
        s3 = Q or R
        kb = s1 and s2 and s3
        print(f"{int(P)} {int(Q)} {int(R)} |  {int(s1)}    {int(s2)}    {int(s3)}  |  {int(kb)}")

display_truth_table()

print("\nEntailment Results:")
print("1. KB entails R:", tt_entails(1))
print("2. KB entails R-> P:", tt_entails(2))
print("3. KB entails Q->R:", tt_entails(3))
```
